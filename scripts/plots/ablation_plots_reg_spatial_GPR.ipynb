```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

folder = "interp_gpr_spatial"

metrics = ["total_plus", "total_minus"]
lrs = np.array([0.003, 0.01, 0.03, 0.1])
kernel_types = ["RBF", "Matern", "Linear"]
ard_num_dimss = np.array([None, 3])
nus = np.array([2.5, 1.5])

TOLERANCE=0.1
```

```python
def error_scatter(df_all, tol, tag):
    
    fig, ax = plt.subplots(figsize=(5, 5))
    
    # Scatter points
    sns.scatterplot(
        data=df_all,
        x="measured",
        y="predicted",
        hue="model",
        alpha=0.7,
        ax=ax
    )
    
    lims = [
        min(df_all["measured"].min(), df_all["predicted"].min()) - 2,
        max(df_all["measured"].max(), df_all["predicted"].max()) + 2
    ]
    
    x = np.linspace(lims[0], lims[1], 100)
    
    # Ideal line
    ax.plot(x, x, "k-", linewidth=2, label="Ideal")
    
    # ±% tolerance
    ax.plot(x, (1.0 + tol) * x, "k--", alpha=0.7)
    ax.plot(x, (1.0 - tol) * x, "k--", alpha=0.7)
    
    ax.axis("equal")
    ax.set_xlim(lims)
    ax.set_ylim(lims)
    
    ax.set_xlabel("Measured [°C]")
    ax.set_ylabel("Predicted [°C]")
    
    ax.set_title(f"Errors {tag}")

    ax.legend(
        loc="upper center",
        bbox_to_anchor=(0.5, -0.12),
        ncol=2,
        frameon=True
    )

    return fig, ax
```

```python
def interval_score(df_all, models, tag):

    df_all["model"] = pd.Categorical(df_all["model"], categories=models, ordered=True)

    fig, ax = plt.subplots(figsize=(7, 5))

    sns.boxplot(
        data=df_all,
        x="model",
        y="interval_score",
        order=models,
        ax=ax
    )

    # Compute medians in the same order as the models
    medians = (
        df_all.groupby("model", observed=False)["interval_score"]
        .median()
        .reindex(models)
    )

    # Add median labels aligned to x positions
    for i, model in enumerate(models):
        y = medians.loc[model]
        ax.text(
            i,
            y,
            f"{y:.2f}",
            ha="center",
            va="bottom",
            fontsize=10,
            fontweight="bold"
        )

    ax.set_ylabel("Interval score")
    ax.set_xlabel("Model")
    ax.set_title(f"Interval score {tag}")

    plt.setp(ax.get_xticklabels(), rotation=45, ha="right")
    fig.tight_layout()

    return fig, ax
```

```python
def coverage(df_all, tag):

    coverage = (
        df_all.groupby("model")["within_pi"]
        .mean()              # True=1, False=0
        .mul(100)            # convert to %
        .reset_index(name="coverage")
    )

    fig, ax = plt.subplots(figsize=(7, 5))
    
    sns.barplot(
        data=coverage,
        x="model",
        y="coverage",
        ax=ax
    )
    
    for container in ax.containers:
        ax.bar_label(container, fmt="%.1f%%", padding=3, fontweight="bold")
    
    
    ax.set_title(f"Coverage {tag}", pad=25)
    ax.set_ylabel("Coverage [%]")
    ax.set_xlabel("Model")
    ax.set_title(f"Coverage {tag}")

    ax.set_ylim(0, 100)
    ax.axhline(95, color="red", linestyle="--", label="Nominal 95%")
    ax.legend()

    plt.setp(ax.get_xticklabels(), rotation=45, ha="right")
    fig.tight_layout()

    return fig, ax
```

```python
folder = "interp_gpr_spatial"


for kernel_type in kernel_types:
    for metric in metrics:
        for nu in nus:

            model_type_prints = []
            for lr in lrs:
                for ard_num_dims in ard_num_dimss:
                    if kernel_type == "Matern":
                        model_type_print = f"{lr}_{kernel_type}_{ard_num_dims}_{nu}"
                        model_type_prints.append(model_type_print)
                    else:
                        nu = None
                        model_type_print = f"{lr}_{kernel_type}_{ard_num_dims}_{nu}"
                        model_type_prints.append(model_type_print)
            print(model_type_prints)
    
            dfs = []
        
            for lr in lrs:
                for ard_num_dims in ard_num_dimss:
                    if kernel_type == "Matern":
                        model_type_print = f"{lr}_{kernel_type}_{ard_num_dims}_{nu}"
                        df_sum = pd.read_csv(f"../../{folder}/ablation_summary_{model_type_print}.csv")
                        # mean_rel_error = df_sum.loc[df_sum["d_type"] == metric, "mean_rel_error"].iloc[0]
                        df = pd.read_csv(f"../../{folder}/ablation_results/{metric}_ablation_{model_type_print}.csv")
                        df["model"] = model_type_print
                        dfs.append(df)
                    else:
                        nu = None
                        model_type_print = f"{lr}_{kernel_type}_{ard_num_dims}_{nu}"
                        df_sum = pd.read_csv(f"../../{folder}/ablation_summary_{model_type_print}.csv")
                        # mean_rel_error = df_sum.loc[df_sum["d_type"] == metric, "mean_rel_error"].iloc[0]
                        df = pd.read_csv(f"../../{folder}/ablation_results/{metric}_ablation_{model_type_print}.csv")
                        df["model"] = model_type_print
                        dfs.append(df)

        
            df_all = pd.concat(dfs, ignore_index=True)
            
            fig, ax = error_scatter(df_all, TOLERANCE, tag=metric)
            save_path = f"../../{folder}/plots/error_scatter_{metric}_{model_type_print}.jpg"
            fig.savefig(save_path, dpi=300, bbox_inches="tight")
            plt.close(fig)

            fig, ax = interval_score(df_all, model_type_prints, tag=metric)
            save_path = f"../../{folder}/plots/interval_score_{metric}_{model_type_print}.jpg"
            fig.savefig(save_path, dpi=300, bbox_inches="tight")
            plt.close(fig)

            fig, ax = coverage(df_all, tag=metric)
            save_path = f"../../{folder}/plots/coverage_{metric}_{model_type_print}.jpg"
            fig.savefig(save_path, dpi=300, bbox_inches="tight")
            plt.close(fig)
```

```python
import pandas as pd
import os

all_results = []

for kernel_type in kernel_types:
    for metric in metrics:
        for nu in nus:

            model_type_prints = []

            for lr in lrs:
                for ard_num_dims in ard_num_dimss:
                    if kernel_type == "Matern":
                        model_type_print = f"{lr}_{kernel_type}_{ard_num_dims}_{nu}"
                    else:
                        model_type_print = f"{lr}_{kernel_type}_{ard_num_dims}_None"

                    model_type_prints.append(model_type_print)

            for model_type_print in model_type_prints:

                # filepath = (f"../../{folder}/ablation_results/"
                #             f"{metric}_ablation_{model_type_print}.csv"
                # )
                filepath = f"../../{folder}/ablation_summary_{model_type_print}.csv"

                df = pd.read_csv(filepath)

                # Store the model information with each result
                df["kernel_type"] = kernel_type
                df["metric"] = metric
                df["nu"] = nu if kernel_type == "Matern" else None
                df["model_type"] = model_type_print

                all_results.append(df)


# Combine all summary files
results = pd.concat(all_results, ignore_index=True)
print(results)
```

```python
best_mean_rel_error = (
    results.loc[
        results.groupby("d_type")["mean_rel_error"].idxmin()
    ]
)

print(best_mean_rel_error[
    ["d_type", "model_type", "mean_rel_error", "pi_coverage", "mean_interval_score"]
])
```

```python
best_interval_score = (
    results.loc[
        results.groupby("d_type")["mean_interval_score"].idxmin()
    ]
)

print(best_interval_score[
    ["d_type", "model_type", "mean_rel_error", "pi_coverage", "mean_interval_score"]
])
```

```python
target_coverage = 0.95

results["coverage_distance"] = (
    results["pi_coverage"] - target_coverage
).abs()

best_coverage = (
    results.loc[
        results.groupby("d_type")["coverage_distance"].idxmin()
    ]
)

print(best_coverage[
    ["d_type", "model_type", "mean_rel_error", "pi_coverage", "mean_interval_score"]
])
```

```python
results["rel_error_rank"] = (
    results.groupby("d_type")["mean_rel_error"]
    .rank(method="min", ascending=True)
)

results["interval_score_rank"] = (
    results.groupby("d_type")["mean_interval_score"]
    .rank(method="min", ascending=True)
)

results["coverage_rank"] = (
    results.groupby("d_type")["coverage_distance"]
    .rank(method="min", ascending=True)
)

results["overall_rank"] = (
    results["rel_error_rank"]
    + results["interval_score_rank"]
    + results["coverage_rank"]
)
```

```python
best_overall = (
    results.loc[
        results.groupby("d_type")["overall_rank"].idxmin()
    ]
)

print(best_overall[
    [
        "d_type",
        "model_type",
        "mean_rel_error",
        "pi_coverage",
        "mean_interval_score",
        "overall_rank",
    ]
])
```
